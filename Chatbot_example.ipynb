```python
while True:
    user = input("Hello, what can i help you bro? ")
    if "weather" in user:
      print("The weather today is sunny and no haze.")
    elif "time" in user:
      print("3PM")
    elif 'bye' in user:
      break
    else:
      print("Sorry, I ain't understand what you saying.")
```

```python
import datetime
from zoneinfo import ZoneInfo

# Define the current time
kl_time = datetime.datetime.now()

while True:
  user = input("Whatsup! What can I help you bruv? ").lower()
  if "weather" in user:
    print("The weather today is sunny and no haze.")
  elif "time" in user:
    kl_time = datetime.datetime.now(ZoneInfo("Asia/Kuala_Lumpur"))
    print("Now is", kl_time.strftime("%H:%M PM"))
  elif "bye" in user:
    print ("Bye! See you again bro.")
    break
  else:
    print("Sorry, I ain't understand what you saying.")
```
