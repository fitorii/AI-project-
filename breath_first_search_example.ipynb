```python
graph = {
    'Home' : ['Bus Stop','Park'],
    'Bus Stop' : ['Office'],
    'Park' : ['Gym'],
    'Gym' : ['Office'],
    'Office' : []

}
visited = []
queue = []
goal = 'Office'

def bfs(visited, graph, node):
    visited.append(node)
    queue.append(node)
    while queue:
        s = queue.pop(0)
        print(s, end="\n")
        for neighbour in graph[s]:
            if neighbour not in visited:
                visited.append(neighbour)
                queue.append(neighbour)
            if goal in visited:
                break

bfs(visited, graph, 'Home')


```
